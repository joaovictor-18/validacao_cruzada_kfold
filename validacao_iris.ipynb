```python
from sklearn.datasets import load_iris
from sklearn.model_selection import StratifiedKFold
from sklearn import svm
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix
```

```python
dataset = load_iris()

x = dataset.data
y = dataset.target
```

```python
modelos = {
    "SVC": svm.SVC(),
    "KNN": KNeighborsClassifier(),
    "Random Forest": RandomForestClassifier(),
    "Decision Tree": DecisionTreeClassifier()
}
```

```python
kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

for nome, model in modelos.items():

    accs = []
    f1s = []

    for idx_treino, idx_teste in kf.split(x, y):

        model.fit(x[idx_treino], y[idx_treino])

        result = model.predict(x[idx_teste])

        accs.append(
            accuracy_score(y[idx_teste], result)
        )

        f1s.append(
            f1_score(y[idx_teste], result, average="macro")
        )

    print(f"{nome}")
    print(f"Accuracy média: {sum(accs) / len(accs):.4f}")
    print(f"F1 médio:       {sum(f1s) / len(f1s):.4f}")
    print()
```
